## Numerical methods
Some equations can be solved analytically. By analytically, we mean getting the exact solution. For example, we can solve the following equation analytically:

$x^2-5x+6=0$

In the above case, $x=2$ and $x=3$.

Now try solving this equation analytically:

$$f(x)=\frac{w_oL}{3\pi^4EI}\left(48L^3cos\left(\frac{\pi}{2L}x\right)-48L^3+3\pi^3Lx^2-\pi^3x^3\right)$$

An exact solution is difficult to find for the above. What we can do is find an approximate solution using numerical methods.

Numerical methods are techniques used to obtain approximate solutions to mathematical problems that may be difficult or impossible to solve exactly.
